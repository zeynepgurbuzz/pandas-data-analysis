# 🚗 Car Market Data Analysis (2010-2020)

## 1. Project Objective
In this project, we analyze a dataset of cars produced between 2010 and 2020. The goal is to understand market trends, price distributions, and the relationship between technical specifications (like engine size) and price.

**Key Steps:**
* Data Cleaning & Preprocessing
* Exploratory Data Analysis (EDA)
* Advanced Visualization (Heatmap, Boxplots)
* Feature Engineering

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("cars_2010_2020.csv")

# Display first 5 rows
print("First 5 rows of the dataset:")
print(df.head())
```

## 2. Data Overview
Understanding the structure of the data, checking for data types and statistical summaries.

```python
print("Dataset Shape (Rows, Columns):")
print(df.shape)

print("\nStatistical Summary:")
print(df.describe())
```

## 3. Data Cleaning & Simulation
Here, we simulate a missing data scenario to demonstrate data handling techniques, filling missing 'Model' values with the mode.

```python
# Simulating missing values
np.random.seed(42)
df.loc[np.random.choice(df.index, 13, replace=False), 'Model'] = np.nan
print("Missing values before handling:")
print(df.isnull().sum())
```

```python
# Filling missing values with Mode
df['Model'] = df['Model'].fillna(df['Model'].mode()[0])
print("\nMissing values after handling:")
print(df.isnull().sum())
```

## 4. Exploratory Data Analysis (EDA)
Visualizing the data to find patterns.

```python
# 1. Price Distribution Histogram
plt.figure(figsize=(10, 6))
sns.histplot(df['Price (USD)'], kde=True, color='blue', bins=30)
plt.title('Distribution of Car Prices (USD)')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
```

```python
# 2. Price vs Fuel Type (Boxplot)
plt.figure(figsize=(10, 6))
sns.boxplot(x='Fuel Type', y='Price (USD)', data=df, palette='Set2')
plt.title('Price Distribution by Fuel Type')
plt.show()
```

## 5. Feature Engineering
Creating a new categorical feature 'Size' and mapping it to numeric values for analysis.

```python
# Adding a random 'Size' feature for demonstration
df['Size'] = np.random.choice(['Small', 'Medium', 'Large'], size=len(df))
size_mapping = {'Small': 1, 'Medium': 2, 'Large': 3}
df['Size_Numeric'] = df['Size'].map(size_mapping)

print(df[['Make', 'Size', 'Size_Numeric']].head())
```

## 6. Correlation Analysis
Checking relationships between numeric variables (Year, Engine Size, Price, Size).

```python
# Select only numeric columns for correlation
numeric_df = df.select_dtypes(include=[np.number])

plt.figure(figsize=(10, 8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

## 7. Conclusion
In this analysis, we cleaned the dataset, handled simulated missing values, and visualized key trends. The heatmap reveals correlations between engine size and price, while the boxplots help identify which fuel types command higher market prices.
