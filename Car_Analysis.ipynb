# 🚗 Car Market Data Analysis (2010-2020)

## 1. Project Objective
In this project, we analyze a dataset of cars produced between 2010 and 2020. We will perform data cleaning, handle missing values, visualize price distributions, and analyze correlations.

**Key Steps:**
* Data Inspection & cleaning
* Handling Missing Data
* Statistical Analysis (Mean, Max, etc.)
* Advanced Visualization (Seaborn Heatmap & Boxplots)
* Feature Engineering (Mapping)

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("cars_2010_2020.csv")

# Display first 15 rows (Your original request)
print("First 15 rows of the dataset:")
print(df.head(15))
```

## 2. Data Overview
Checking data structure, column names, and types.

```python
print("\nTotal number of rows and columns:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

## 3. Statistical Summary & Fuel Distribution
Analyzing the basic statistics and counting fuel types.

```python
print("Statistical informations:")
print(df.describe())

fuel_type_counts = df['Fuel Type'].value_counts()
print("\nFuel Type Distribution:")
print(fuel_type_counts)
```

## 4. Handling Missing Data (Simulation)
Simulating missing values in 'Model' column and filling them with Mode.

```python
print("Missing Datas (Before Simulation):")
print(df.isnull().sum())

# Simulating missing values
np.random.seed(42)
df.loc[np.random.choice(df.index, 13, replace=False), 'Model'] = np.nan
print("\nMissing datas (After Simulation):")
print(df.isnull().sum())
```

```python
# Filling missing values with Mode
df['Model'] = df['Model'].fillna(df['Model'].mode()[0])
print("\nAfter filling in the missing data:")
print(df.isnull().sum())
```

## 5. Specific Statistical Insights
Calculating average year and maximum year as requested.

```python
print("Average of 'Year' column:")
print(df['Year'].mean())

print("\nHighest value of 'Year' column:")
print(df['Year'].max())
```

## 6. Exploratory Data Analysis (EDA)
Visualizing Price distribution and Fuel Type impact.

```python
# 1. Price Distribution Histogram
print("Histogram of Price column:")
plt.figure(figsize=(10, 6))
sns.histplot(df['Price (USD)'], kde=True, color='blue', bins=30)
plt.title('Distribution of Car Prices (USD)')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
```

```python
# 2. Price vs Fuel Type (Boxplot)
plt.figure(figsize=(10, 6))
sns.boxplot(x='Fuel Type', y='Price (USD)', data=df, palette='Set2')
plt.title('Price Distribution by Fuel Type')
plt.show()
```

## 7. Data Selection & Cleaning
Selecting specific columns and dropping the 'Model' column for further analysis.

```python
selected_columns = df[['Make', 'Model', 'Price (USD)']]
print(selected_columns.head(10))

# Dropping Model Column
df = df.drop('Model', axis=1)
print("\nDataframe after dropping 'Model':")
print(df.head(10))
```

## 8. Feature Engineering (Mapping)
Adding a categorical 'Size' feature and mapping it to numeric values.

```python
# Adding a random 'Size' feature
np.random.seed(42)
df['Size'] = np.random.choice(['Small', 'Medium', 'Large'], size=len(df))

# Mapping Size to Numeric
df['Size'] = df['Size'].map({'Small': 1, 'Medium': 2, 'Large': 3})
print(df.head(10))
```

## 9. Correlation Analysis
Checking relationships between numeric variables (Year, Engine Size, Price, Size).

```python
# Select only numeric columns for correlation
numeric_df = df.select_dtypes(include=[np.number])

plt.figure(figsize=(10, 8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

## 10. Conclusion
In this comprehensive analysis, we successfully cleaned the data, handled missing values, and performed feature engineering. The visualization and correlation matrix provide insights into factors affecting car prices.
